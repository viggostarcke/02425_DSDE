## Question 3

```python
import numpy as np
import matplotlib.pyplot as plt

N = 10_000
n = 10

X = np.random.uniform(0,1, N)
Y = np.random.uniform(0,1, N)
Z = X + Y

x_plot = np.zeros(n)
z_mean_plot = np.zeros(n)

for i in range(1,n+1):
    mask = (X >= (i - 1)/n) & (X < i/n)
    filtered_Z = Z[mask]
    x_plot[i-1] = i / n
    z_mean_plot[i-1] = np.mean(filtered_Z)
    
bin_centers = x_plot - 1 / (2 * n)

plt.figure(figsize=(8, 5))
plt.plot(bin_centers, z_mean_plot, marker='o', linestyle='-', color='blue', label='Empirical E{Z | X}')

plt.plot(bin_centers, bin_centers + 0.5, linestyle='--', color='red', label='Theoretical E{Z | X}')

plt.xlabel('X (Upper bound of bin i/n)')
plt.ylabel('Empirical E{Z | X}')
plt.title('Conditional Expectation of Z given X')
plt.legend()
plt.grid(True)
plt.show()
```

## Question 5

```python
N = 10_000
n = 10

X = np.random.uniform(0,1, N)
Y = np.random.uniform(0,1, N)
Z = X + Y

z_plot = np.zeros(n)
x_mean_plot = np.zeros(n)

for i in range(1,n+1):
    mask = (Z >= ((i*2) - 2)/n) & (Z < (i*2)/n)
    filtered_X = X[mask]
    z_plot[i-1] = (i*2)/n
    x_mean_plot[i-1] = np.mean(filtered_X)
    
bin_centers = z_plot - (1 / n)

plt.figure(figsize=(8, 5))
plt.plot(bin_centers, x_mean_plot, marker='o', linestyle='-', color='blue', label='Empirical E{X | Z}')

# The theoretical line h(Z) = Z / 2
# We adjust the x-axis to the center of the bins for the theoretical line
# Bin width is 2/n, so we subtract 1/n from the upper bound to find the center
plt.plot(bin_centers, bin_centers / 2, linestyle='--', color='red', label='Theoretical E{X | Z} = Z/2')

plt.xlabel('Z (Upper bound of diagonal bin)')
plt.ylabel('Empirical E{X | Z}')
plt.title('Conditional Expectation of X given Z')
plt.legend()
plt.grid(True)
plt.show()
```

## Question 7

```python
N = 10_000
n = 10

X = np.random.uniform(0,1, N)
Y = np.random.uniform(0,1, N)
Z = X + Y

z_plot = np.zeros(n)
x_var_plot = np.zeros(n)

for i in range(1,n+1):
    mask = (Z >= ((i*2) - 2)/n) & (Z < (i*2)/n)
    filtered_X = X[mask]
    z_plot[i-1] = (i*2)/n
    x_var_plot[i-1] = np.var(filtered_X)
    
bin_centers = z_plot - (1 / n)

plt.figure(figsize=(8, 5))
plt.plot(bin_centers, x_var_plot, marker='o', linestyle='-', color='blue', label='Empirical V{X | Z}')

theoretical_var = np.where(bin_centers <= 1, (bin_centers**2) / 12, ((2 - bin_centers)**2) / 12)

plt.plot(bin_centers, theoretical_var, linestyle='--', color='red', label='Theoretical V{X | Z}')
plt.xlabel('Z (Upper bound of diagonal bin)')
plt.ylabel('Variance of X')
plt.title('Conditional Variance of X given Z')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print(np.var(X))
print(np.mean(np.where(Z <= 1, (Z**2)/12, ((2 - Z)**2)/12)))
print(np.var(Z/2))
```
